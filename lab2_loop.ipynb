```python
n = 10
while n > 0:
    print(n)
    n = n - 1
```

```python
# 구구단
i = 1
while i <= 10:
    j = 1
    while j <= 10:
        print(i * j, end='\t')
        j += 1
    print()
    i += 1
```

```python
# range: generates a sequence of integers
for i in range(1, 11, 2):
    print(i)
```

```python
# [a, b)로 처리하는 이유:
# 구간을 쪼개서 처리해야 할 때 편하다.
for i in range(1, 5):
    print(i)
for i in range(5, 10):
    print(i)
for i in range(10, 15):
    print(i)
```

```python
for i in range(1, 11):
    for j in range(1, 11):
        print(i * j, end='\t')
    print()
```

```python
# Fibonacci

r = int(input('Enter a number (n > 0):'))
n_2=0
n_1=1
print(n_2)
print(n_1)
# Pythonic
for _ in range(2, r + 1):
    n_2, n_1 = n_1, n_1 + n_2
    print(n_1)
```

```python
n = 1
while 42:
    if n > 10:
        break
    m = 1
    while 42:
        if m > 10:
            break
        print(n * m, end='\t')
        m += 1
    print()
    n += 1
```
