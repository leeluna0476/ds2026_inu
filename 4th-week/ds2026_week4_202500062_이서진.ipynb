```python
# 3
def check_UTC(x):
    """Check current UTC
    Arg:
        x = input is string
            want to see time part (etc : year, month, ...)
    Returns:
        output is time
        have only "year", "month", "day", "hour", "min", "sec" Output!
    """
    units = [ 'year', 'month', 'day', 'hour', 'min', 'sec' ]
    if x in units:
        for i in range(len(units)):
            if units[i] == x:
                break;
        print(time.gmtime()[i])
        print(x)
    else:
        print('Wrong input! Please try different input!')

time_part = input('Enter want to see time part: ')
check_UTC(time_part)
```

```python
# 4
def recursion_power(n):
    """Sum 1 to n then, each number is squared!
    Args:
        n:
            input is int and destination number
    Returns:
        Output is summation 1^2 to n^2
    """
    if n == 1:
        return 1
    else:
        return n ** 2 + recursion_power(n - 1)

print(recursion_power(4))
```

```python
# 5
def loop_power(n):
    """Sum 1 to n then, each number is squared!
    Args:
        n:
            input is int and destination number
    Returns:
        Output is summation 1^2 to n^2
    """
    r = 1
    for i in range(2, n + 1):
        r += i ** 2
    return r

print(loop_power(5))
```

```python
# 6
def check_func_time(func, num):
    """Measures the time of 'func' executed.
    Args:
        func:
            the function to be executed
        num:
            the argument to be given to 'func'
    Returns:
        the measured time of 'func' executed
    """
    start_point = time.perf_counter_ns()
    result = func(num)
    print(result)
    end_point = time.perf_counter_ns()
    return end_point - start_point

func_work_time = check_func_time(loop_power, 5)
print(func_work_time)
help(check_func_time)
```

```python
# 7
lps = loop_power
print(lps(5))

# loop_power라는 이름(변수)을 가진 오브젝트(타입: function)에 lps라는 또다른 이름을 붙였다.
# 이로써 해당 오브젝트에 lps로 접근이 가능해졌고, loop_power(5)와 똑같이 실행이 가능하다.
```

```python
# 8
# 8-1
power_sum = [ recursion_power, loop_power ]
# 8-2
iter = 5
for f in power_sum:
    result_time = 0.0
    for _ in range(iter):
        x = check_func_time(f, 5)
        result_time += x
    result_time /= iter
    print(f'{f.__name__}: {result_time}')
```

```python
# 10
def unfrog(x):
    if type(x) == int:
        print(chr(x))
    else:
        print(ord(x))

unfrog(77)
unfrog('M')
```

```python
# 11
def ununfrog(x):
    if type(x) == int:
        print(chr(x))
    else:
        d = ord(x)
        print(bin(d))
        print(oct(d)) # 문제 오타
        print(hex(d))

ununfrog('B')
```
