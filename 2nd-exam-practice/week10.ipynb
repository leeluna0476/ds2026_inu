```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
help(plt.legend)
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots()
ax.plot(x, y1, label='y=sin(x)')
ax.plot(x, y2, label='y=cos(x)')
ax.set(xlabel='x', ylabel='y')
ax.set_title('sin/cos')
ax.legend(loc='upper right')

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].plot(x, y1)
axes[0].set(xlabel='x', ylabel='y')
axes[0].set_title('y=sin(x)')
axes[0].grid(axis='both')

axes[1].plot(x, y2)
axes[1].set(xlabel='x', ylabel='y')
axes[1].set_title('y=cos(x)')
axes[1].grid(axis='both')
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)
```

```python
x = np.linspace(-10, 10, 100)
y1 = sigmoid(x)
y2 = relu(x)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].plot(x, y1, label='sigmoid(x)')
axes[1].plot(x, y2, label='relu(x)')

axes[0].grid(axis='both')
axes[1].grid(axis='both')

axes[0].set_yticks(np.arange(0, 1.1, 0.1))
axes[0].set_xticks(np.linspace(-10, 10, 5))

axes[1].set_yticks(np.arange(0, 12, 2))
axes[1].set_xticks(np.linspace(-10, 10, 5))

axes[0].set(xlabel='x', ylabel='y')
axes[1].set(xlabel='x', ylabel='y')

axes[0].legend()
axes[1].legend()

axes[0].set_title('y=sigmoid(x)')
axes[1].set_title('y=relu(x)')

plt.show()
```
