## Analysis with plots

```python
age = df['Age']
ax = age.hist(bins=20, edgecolor='k')
ax.set(xlabel='age', ylabel='counts')
ax.set_title('Age Distribution')
```

```python
income = df['Income']
ax = income.hist(bins=20, edgecolor='k')
ax.set(xlabel='income', ylabel='counts')
ax.set_title('Income Distribution')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10,5))
ax1 = age.hist(ax=axes[0], bins=20, edgecolor='k')
ax1.set(xlabel='age', ylabel='counts')
ax1.set_title("Age Distribution")

ax2 = income.hist(ax=axes[1], bins=20, edgecolor='k')
ax2.set(xlabel='income', ylabel='counts')
ax2.set_title("Income Distribution")
```

```python
females = df['Gender'] == 'female'
df[females]
```

```python
print(f'female={len(df[females]) / len(df) * 100:.2f}%')
```

```python
males = df['Gender'] == 'male'
print(f'male={len(df[males]) / len(df) * 100:.2f}%')
```

```python
df_female = df[females]
df_male = df[males]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axf = df_female['Age'].hist(ax=axes[0], bins=20, edgecolor='k', color='m')
axf.set_title('Female Age Distribution')
axf.set(xlabel='Age', ylabel='Counts')
axm = df_male['Age'].hist(ax=axes[1], bins=20, edgecolor='k', color='c')
axm.set_title('Male Age Distribution')
axm.set(xlabel='Age', ylabel='Counts')
```

```python
fig, ax = plt.subplots()
df_male['Age'].hist(ax=ax, bins=20, edgecolor='k', color='c', alpha=0.5, label='male')
ax = df_female['Age'].hist(ax=ax, bins=20, edgecolor='k', color='m', alpha=0.7, label='female')

ax.set(xlabel='Age', ylabel='Count')
ax.set_title("Age Distribution")
ax.legend()
```

## Statistics

```python
# mean of learning hours
print(f'Average hours learning: {df['HoursLearning'].mean():.2f}h')
print(f'Median hours learning: {df['HoursLearning'].median():.2f}h')
```

```python
q2 = df['HoursLearning'].quantile(0.5)
q3 = df['HoursLearning'].quantile(0.75)
q1 = df['HoursLearning'].quantile(0.25)

iqr = q3 - q1
upper_bound = q3 + 1.5 * iqr
lower_bound = q1 - 1.5 * iqr

print(f'q1={q1}, q2={q2}, q3={q3}, iqr={iqr}, bounds=[{upper_bound}, {lower_bound}]')
```
