```python
# 10
f = open('Atoz.txt', 'w', encoding='utf-8')
for a in range(ord('a'), ord('z') + 1):
    f.write(chr(a))
f.close()
```

```python
# 11
f = open('Atoz.txt', 'r', encoding='utf-8')
data1 = f.read()
f.close()
alphadict = {}
for a in data1:
    alphadict[a] = 0
print(alphadict)
```

```python
# 12
# 6번에서 만든 lowers 그대로 사용
for a in range(ord('a'), ord('z') + 1):
    c = chr(a)
    alphadict[c] = lowers.count(c)
print(alphadict)
```

```python
# 13
most_used = max(alphadict.items(), key=lambda x: x[1])
least_used = min(alphadict.items(), key=lambda x: x[1])
print(f'the mostly used alphabet is {most_used[0]} ({most_used[1]} times)')
print(f'the leastly used alphabet is {least_used[0]} ({least_used[1]} times)')
# 13-1
most_used1 = ['?', 1]
least_used1 = ['?', most_used[1]]
most_used2 = ['?', 0]
least_used2 = ['?', most_used[1]]
for k, v in alphadict.items():
    if v < most_used[1]:
        if k == 'a':
            print(f'a={v}, {v > most_used1[1]}')
        if v >= most_used1[1]:
            if most_used1[1] > most_used2[1]:
                most_used2[0] = most_used1[0]
                most_used2[1] = most_used1[1]
            most_used1[0] = k
            most_used1[1] = v
        elif v >= most_used2[1]:
            most_used2[0] = k
            most_used2[1] = v
    if v > least_used[1]:
        if v <= least_used1[1]:
            if least_used1[1] < least_used2[1]:
                least_used2[0] = least_used1[0]
                least_used2[1] = least_used1[1]
            least_used1[0] = k
            least_used1[1] = v
        elif v <= least_used2[1]:
            least_used2[0] = k
            least_used2[1] = v

print(f'the leastly used alphabet is {least_used[0]} ({least_used[1]} times)')
print(f'the mostly used alphabet is {most_used[0]} ({most_used[1]} times)')
print(f'the 2nd leastly used alphabet is {least_used1[0]} ({least_used1[1]} times)')
print(f'the 2nd mostly used alphabet is {most_used1[0]} ({most_used1[1]} times)')
print(f'the 3rd leastly used alphabet is {least_used2[0]} ({least_used2[1]} times)')
print(f'the 3rd mostly used alphabet is {most_used2[0]} ({most_used2[1]} times)')
```

```python
# 14
f = open('1to1000.txt', 'w', encoding='utf-8')
s = str([n for n in range(1, 1001) if n & 1]) + '\n'
f.write(s)
f.close()
```
