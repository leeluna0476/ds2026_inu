```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = Image.open('cat_large.jpg')
nimage = np.array(img)
plt.imshow(nimage)
```

```python
butt = nimage[500:1000, 1100:1500, :]
plt.imshow(butt)
```

```python
plt.imshow(butt[-1:0:-1, :, :])
```

```python
def downscale(img, scale_factor=2):
    cropy = img.shape[0] // scale_factor * scale_factor
    cropx = img.shape[1] // scale_factor * scale_factor
    print(cropy, cropx)
    crop = img[0:cropy, 0:cropx, :]
    crop = crop.reshape((cropy // scale_factor, scale_factor,
                        cropx // scale_factor, scale_factor, -1))
    crop = crop.sum(axis=(1, 3)) // (scale_factor * scale_factor)
    return crop
```

```python
downscaled = downscale(butt, 3)
plt.imshow(downscaled)
```
