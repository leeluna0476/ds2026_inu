```python
# lab
def sum_nested_list(l):
    r = 0
    for e in l:
        if isinstance(e, list):
            r += sum_nested_list(e)
        else:
            r += e
    return r
```

# Class
- class comes with state and behaviour
- -> attributes(values) and methods(functions used to manipulate attributes)
- attributes are accessed via methods -> 'encapsulation'

```python
# modeling a 2-D vector point
# class의 이름은 대문자로 시작해야 한다
class Point:
    # self: instance가 만들어지고 나서 method를 호출할 때, 그 object 자신을 가리킨다
    # __init__: instance 생성 시 호출된다. 항상 가지고 있어야 할 attribute 생성 가능.
    # dunder(double underbars) methods: 특별한 의미를 가지는 메소드.
    def __init__(self, x=.0, y=.0):
        self.__x = x
        self.__y = y
    
    def __str__(self):
        return f'({self.__x:.4f}, {self.__y:.4f})'
    
    def __repr__(self):
        return f'({self.__x:.4f}, {self.__y:.4f})'
    
    # __len__은 항상 정수만 return한다
    def __len__(self):
        return int(self.norm())
    
    def __add__(self, p2):
        return Point(self.__x + p2.__x, self.__y + p2.__y)
    
    def __sub__(self, p2):
        return Point(self.__x - p2.__x, self.__y - p2.__y)
    
    # element-wise vector multiplication
    def __mul__(self, p2):
        return Point(self.__x * p2.__x, self.__y * p2.__y)
    
    def __getitem__(self, key):
        if key == 1:
            return self.__x
        elif key == 2:
            return self.__y
        else:
            raise IndexError(f'1 <= key <= 2')

    def norm(self):
        return (self.__x ** 2 + self.__y ** 2) ** 0.5
    
    def distance(self, p2):
        return ((self.__x - p2.__x) ** 2 + (self.__y - p2.__y) ** 2) ** 0.5

p1 = Point(1, 2)
# add attributes
# 자유롭게 추가할 수 있다
# 그러나 모두 동일한 attribute를 가지도록 설계하는 것이 낫다.
# 값에 직접 접근하는 것은 좋지 않다.
# p1.x = 3
# p1.y = 4

p2 = Point(3, 4)
p3 = Point()

points = [p1, p2, p3]
for p in points:
    print(f'norm: {p.norm():.4f}')
    print(f'len: {len(p)}')

print(f'||p1-p2||={p1.distance(p2):.4f}')
print(f'||p2-p1||={p2.distance(p1):.4f}')

p4 = p1 + p2
print(f'p1+p2={p4}')

p5 = p1 - p2
print(f'p1-p2={p5}')

p6 = p1 * p2
print(f'p1*p2={p6}')

print(f'p1[1]={p1[1]}')
print(f'Invalid: p1[3]={p1[3]}')
```
