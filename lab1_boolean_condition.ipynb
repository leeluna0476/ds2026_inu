```python
x = int(input('Enter a number: '))
# indentation = space 4
if x > 0:
    print('x is positive')
elif x < 0:
    print('x is negative')
else:
    print('x is zero')
```

```python
height = float(input('What is your height (cm)? ')) / 100.
weight = float(input('What is your weight (kg)? '))
bmi = weight / (height ** 2)
print(f'Your BMI is {bmi:.2f}')
if bmi < 18:
    print('Underweight')
elif bmi < 25:
    print('normal weight')
elif bmi < 27: # 25 <= bmi < 27 이렇게 수학적으로 나타낼 수 있다.
    print('overweight')
elif bmi < 30:
    print('moderate obese')
elif bmi < 40:
    print('severe obese')
else:
    print('very severe obese')
```
