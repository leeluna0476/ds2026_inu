# Sets
- **unordered** collection of **distinct** objects
- *mutable*, but *the items are unchangable*
  - only immutable objects can be inserted into a set

```python
[] # an empty list
() # an empty tuple: unnecessary
(10,) # a tuple with one element (interpreted as an int object without comma)
set() # an empty set
{1, 2, 3} # a non-empty set
```

```python
low = {0, 1, 2, 3, 4, 2}
low # 중복 없음
```

```python
odds = [1, 3, 5, 7, 9, 5]
odds = set(odds) # 중복을 제거하고 싶을 때 유용하다
odds.remove(5)
odds.add(11)
odds
```

## Set operations
- union
- intersection
- difference
- ...

```python
low = {0, 1, 2, 3, 4, 2}
odds = {1, 3, 5, 7, 9, 5}
ten = set(range(10))

# print(low.intersection(odds))
# print(low.union(odds))

print(f'intersection: {low & odds}')
print(f'union: {low | odds}')
print(f'difference: {low - odds}')

# print(low.issubset(ten))
# print(low.issubset(odds))
print(f'subset: {low <= ten}')
print(f'proper subset: {low < ten}')

low.clear()
print(low)
```

- sets are iterable objects
- every object has its unique id

```python
# myset = {1, 2, 'hello', (1, 2, 3), [2, 3]} # list는 mutable이라 set의 element가 될 수 없다.
# list -> unhashable
myset = {1, 2, 'hello', (1, 2, 3)}
for x in myset:
    print(x)
```

```python
# x and y are different objects
x = (1, 2, 3)
y = (1, 2, 3)
print(id(x))
print(id(y))
print(x is y) # object id가 같으면 true
print(x == y) # element가 같으면 true
# element가 같은지는 어떻게 판별하나?
```

```python
# dir(x)
# hash: object의 값을 수학적인 함수에 넣어서 고유한 값을 계산한 것.
# 초기에 hash 저장해두면 중복 검사할 때마다 해시만 비교하면 된다.
# mutable object는 고유한 해시를 계산할 수 없기 때문에 unhashable하다. (list, set, dict, ..)
# unhashable하다면 중복 검사를 할 수 없기 때문에 set에 넣을 수 없다.
print('x.__hash__:', x.__hash__())
print('y.__hash__:', y.__hash__())
```

# Dictionaries
- 특징: 단어와 그에 대한 설명이 있다.
- 단어: key
- 설명: value
- key를 보고 value를 찾는 자료구조
- **mutable**
- **ordered** (Since Python 3.7)
- **iterable**

**similar data structures**
- map, hash, associative array, key/value pairs
- turboquant kv cache..

```python
d = {} # an empty dict
type(d)
```

```python
# key는 hashable해야 한다.
# value는 상관없다.
birthday = {'Newton': 1642, 'Darwin': 1809}
print(birthday)
print(birthday['Darwin']) # index by key
```

```python
birthday['Darwin'] = 2004 # mutable
birthday['Turing'] = 1912
del birthday['Darwin']
print(birthday)
```

```python
birthday = {
    'Newton': 1642,
    'Darwin': 1809,
    'Turing': 1912,
    'Bong': 2004
}

# iterator를 불러오면 key값만 나온다.
# for x in birthday:
#     print(x, birthday[x])

# key, value를 동시에 불러오고 싶다면?
for k, v in birthday.items():
    print(k, v)
```

```python
# lab: histogram construction

birds = ['tern', 'goose', 'goose', 'hawk', 'tern', 'goose', 'tern']
# in keyword 사용 가능
observations = {}

for bird in birds:
    if bird not in observations:
        observations[bird] = 1
    else:
        observations[bird] += 1

print(observations)
```

```python
print(observations.keys())
print(observations.values())
print(observations.items())
```

```python
observations.get('dove', -1) # dove가 없다면 -1을 반환하라
```

```python
# lab using get
birds = ['tern', 'goose', 'goose', 'hawk', 'tern', 'goose', 'tern']
observations = {}

for bird in birds:
    observations[bird] = observations.get(bird, 0) + 1

print(observations)
```
