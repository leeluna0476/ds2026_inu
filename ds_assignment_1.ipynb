```python
# (x1, y1) = bottom-left corner of a rectangle
# (x2, y2) = top-right corner of a rectangle

# case #1
ax1, ay1, ax2, ay2 = 1, 2, 4, 5
bx1, by1, bx2, by2 = 2, 1, 5, 4

if bx1 < ax2 and bx2 > ax1 and by1 < ay2 and by2 > ay1:
    overlap_width = min(ax2, bx2) - max(ax1, bx1)
    overlap_height = min(ay2, by2) - max(ay1, by1)
    overlap_area = overlap_width * overlap_height
    print('overlap!')
    print('area:', overlap_area)
else:
    print('no overlap!')
```
