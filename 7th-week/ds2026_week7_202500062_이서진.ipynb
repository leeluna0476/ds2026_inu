```python
# 2
class Classroom:
    def __init__(self, grade, name):
        self.grade = grade
        if (isinstance(name, str)):
            self.name = [name]
        else:
            self.name = name

    def __str__(self):
        return f'{self.grade}_Grade : {self.name}'
    
    def __len__(self):
        return len(self.name)
    
    def add_student(self, grade, name):
        if grade == self.grade:
            if isinstance(name, str):
                self.name.append(name)
            else:
                self.name.extend(name)
        else:
            print('Invalid Input')

f_grade = Classroom(2, 'seojilee')
print(f_grade)

f_grade.add_student(3, 'bongjina')
print(f_grade)

f_grade.add_student(2, 'bongjina')
print(f_grade)

f_grade.add_student(2, ['jina', 'bong'])
print(f_grade)

print(len(f_grade))
```

```python
# 4
info = {
    'high_school': {
        '1': ['sunwoo', 'Lee'],
        '2': 'kim',
        '3': 'park'
    },
    'middle_school': {
        '2': ['Andrew', 'Yoshua']
    }
}

students = {}
for k1, v1 in info.items():
    students[k1] = [Classroom(int(k2), v2) for k2, v2 in v1.items()]

for i in students:
    print(i)
    for j in students[i]:
        print(j)
```

```python
# 5
high_school_num = [{f'{c.grade}grade': len(c)} for c in students['high_school']]
middle_school_num = [{f'{c.grade}grade': len(c)} for c in students['middle_school']]
print(high_school_num)
print(middle_school_num)
```

```python
# 6
h_sum = sum([sum(c.values()) for c in high_school_num])
m_sum = sum([sum(c.values()) for c in middle_school_num])
print(f'high_school : {h_sum}')
print(f'middle_school : {m_sum}')
```
