```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
pil_image = Image.open('cat_large.jpg')
np_image = np.array(pil_image)
plt.imshow(np_image)
```

```python
def better_downscale(np_image, scale_factor=1):
    sh0 = np_image.shape[0] // scale_factor * scale_factor
    sh1 = np_image.shape[1] // scale_factor * scale_factor
    sliced = np_image[:sh0, :sh1, :]
    ax0 = sliced.shape[0] // scale_factor
    ax1 = sliced.shape[1] // scale_factor
    split = sliced.reshape((ax0, scale_factor, ax1, scale_factor, -1))
    return split.sum(axis=(1, 3)) // (split.shape[1] * split.shape[3])
```

```python
downscaled_image = better_downscale(np_image,10)
plt.imshow(downscaled_image)
```
