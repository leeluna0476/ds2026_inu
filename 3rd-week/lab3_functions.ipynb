## Parameter vs Argument
- Parameter: 함수 정의할 때 명명하는 변수.
- Argument: parameter에 실제로 할당하는 값.

```python
def double(x):
    """
    Calculate a double of 'x'.
    Args:
        x: int
    Returns:
        twice of x
    """
    return 2 * x
```

```python
print(double(2))
# 용도에 맞지 않게 사용해도 에러가 발생하진 않는다.
print(double(2.2))
print(double('bong'))
print(double([1, 2, 3, 'h']))
```

```python
def print_square(size: int):
    """
    Prints a square with '#' symbols.
    Args:
        size: defines the length of sides
    """
    for _ in range(size):
        for _ in range(size):
            print('#', end='')
        print()
```

```python
print_square(3)
print_square(5)
```

## Function의 중요한 특징
- Function 또한 object다.
  - 데이터가 아닌 **코드** object.
- 함수를 호출하면 스택 메모리 맨 위에 함수의 스택 프레임이 생긴다. 함수 내에서 사용되는 변수들이 여기에 push된다. 함수가 반환되면 pop되어 없어진다. 이를 local variable이라고 한다.
- 호출된 함수는 자기 스택 프레임 내의 변수와 전역변수만 볼 수 있다.

```python
def factorial(x):
    fact = 1
    while x > 1:
        fact *= x
        x -= 1
    return fact
```

```python
def factorial_r(x):
    """
    Returns:
        x >= 0: factorial of x
        otherwise: -1, which refers an error
    """
    if x == 0 or x == 1:
        return 1
    elif x > 1:
        return x * factorial_r(x - 1)
    else:
        return -1
```

```python
factorial_r(3)
```

```python
factorial_r(-10)
```

```python
def sum_to_num(num):
    # return (1 + num) * num // 2
    if num == 1:
        return num
    return num + sum_to_num(num - 1)
```

```python
sum_to_num(100)
```

```python
sum_to_num(num=10)
```

- function의 parameter에는 default 값을 줄 수 있다.

```python
# def my_func(name='Info', country) 이런 식으로는 안된다.
# default값이 있는 파라미터는 항상 맨 뒤에 와야 한다.
def my_func(name, country='Norway', age=0):
    print(name, 'is from', country, 'and is', age, 'year(s) old')
```

```python
my_func('seojilee')
my_func('seojilee', country='Korea')
my_func('seojilee', 'Korea', 21)
my_func('seojilee', age=21, country='Korea') # 순서 바꿔서 쓸 거면 parameter 이름 명시해야함.
```

```python
fp = my_func # my_func가 가리키는 object에 새로운 이름을 붙여준다.
fp('bongjina')
```
