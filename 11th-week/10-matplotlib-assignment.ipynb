```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)
```

```python
x = np.linspace(-10, 10, 100)
y = [sigmoid(x), relu(x)]
labels = ['sigmoid(x)', 'relu(x)']

fig, ax = plt.subplots(1, 2, figsize=(15,5))
for i in range(2):
    ax[i].plot(x, y[i], label=labels[i])
    ax[i].grid(axis='both')
    ax[i].set(xlabel='x', ylabel='y')
    ax[i].set_xticks(np.linspace(-10, 10, 5))
    ax[i].set_yticks(np.linspace(0, y[i][-1], 10 // (i + 1) + 1))
    ax[i].legend()
```
