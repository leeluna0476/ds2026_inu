```python
class Classroom:
    def __init__(self, grade, name):
        self.grade = grade
        if isinstance(name, str):
            self.name = [name]
        else:
            self.name = name

    def __str__(self):
        return f'{self.grade}_Grade : {self.name}'
    
    def add_student(self, grade, name):
        if grade != self.grade:
            print('Invalid Input')
        elif isinstance(name, str):
            self.name.append(name)
        else:
            self.name.extend(name)
    
    def __len__(self):
        return len(self.name)

f_grade = Classroom(1, ['sunwoo', 'Lee'])
print(f_grade)

f_grade.add_student(2, ['Kim', 'Park'])
print(f_grade)

f_grade.add_student(1, ['Kim', 'Park'])
print(f_grade)

print(len(f_grade))
```

```python
info = {
    'high_school': {
        '1': ['sunwoo', 'Lee'],
        '2': 'kim',
        '3': 'park'
    },
    'middle_school': {
        '2': ['Andrew', 'Yoshua']
    }
}

school_data = {}
for k,v in info.items():
    school_data[k] = []
    for k2, v2 in v.items():
        school_data[k].append(Classroom(int(k2), v2))

for i in school_data:
    print(i)
    for j in school_data[i]:
        print(j)
```

```python
high_school_num = [{str(v.grade) + 'grade': len(v)} for v in school_data['high_school']]
middle_school_num = [{str(v.grade) + 'grade': len(v)} for v in school_data['middle_school']]

print(high_school_num)
print(middle_school_num)
```

```python
h_sum = sum([sum(v.values()) for v in high_school_num])
m_sum = sum([sum(v.values()) for v in middle_school_num])

print(f'high_school : {h_sum}')
print(f'middle_school : {m_sum}')
```
