# NumPy
- numpy가 다루는 데이터: n-dimensional arrays로 저장된다.
- n-dimensional arrays: also called Tensors
- scalars (0D, rank 0 tensor)
- vectors (1D, rank 1 tensor)
- matrices (2d, rank 2 tensor)
- 3d and above (rank 3 tensor and above)

```python
import numpy as np
```

```python
# scalar
# rank 0 tensor
# 0D array
a = np.array(42)
print(a)
print(type(a))
print(a.ndim)
print(a.dtype) # array의 data type
print(a.shape)
```

```python
# vector
# 1D array
# rank 1 tensor
# audio signals, sensor values, ...
a = np.array([1, 2, 3]) # list를 vector로...
print(a)
print(type(a))
print(a.ndim)
print(a.dtype)
print(a.shape) # vector의 길이 (element #)
```

```python
# matrix
# 2D array
# rank 2 tensor
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
print(type(a))
print(a.ndim)
print(a.dtype)
print(a.shape) # (column len, row len) axis1 길이=2, axis2 길이=3
               # 2x3
```

```python
# 3D array
# rank 3 tensor
# matrix가 여러 개
a = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])
print(a)
print(type(a))
print(a.ndim)
print(a.dtype)
print(a.shape) # 2x2x3
```

# list vs ndarray

```python
import time

a = list(range(10000))
b = list(range(10000, 20000))
# c = [a[i] + b[i] for i in range(10000)]

start = time.time()
for _ in range(10000):
    c = [n1 + n2 for n1, n2 in zip(a, b)]
end = time.time()
tlist = end - start
print(f'list={tlist:.6f}')
# print(c[:10])
# help(zip)
```

```python
na = np.array(a)
nb = np.array(b)

start = time.time()
# 병렬로 처리해서 빠르다.
# GPU로 한번에 여러 element 연산.
# GPU는 내부에 코어가 수만개씩 들어있다. -> 한번에 수만개씩 연산.
# CPU로만 계산해도 병렬 처리가 가능하다. -> SIMD instruction. 이를 위한 레지스터가 따로 있다.
# 근데 numpy는 GPU쓰는 건 아니고 SIMD를 쓴다.
for _ in range(10000):
    nc = na + nb
end = time.time()
tndarray = end - start
print(f'ndarray={tndarray:.6f}')
# print(nc[:10])
```

```python
tlist / tndarray
```

```python
# indexing
a = np.arange(20)
print(a)
print(a[3])
print(a[:3]) # slicing
print(a[3:7])
print(a[7:])

b = a[:5] # python은 copy인데, numpy는 아니다.
print(b)
b[0] = 100
print(a)

# view: slicing하면 그 부분으로 연결만 해준다. 주로 데이터가 아주 크기 때문에 오버헤드 줄이기 위함.

c = a.copy()
c[0] = 200 # real copy
print(a)
```

```python
# fancy indexing
a = np.random.randint(0, 10, size=(10,))
print(a)

# fancy indexing은 copy를 한다.
b = a[[0, 4, 2]]
b[0] = 100
print(f'a={a}')
print(f'b={b}')
```

```python
# boolean indexing
a = np.random.uniform(-1, 1, size=(10,))
print(a)

b = a < 0
print(b)

print(a[b]) # 인덱스가 true인 부분만 가져온다.
print(a[a > 0])

c = [a < 0] # fancy indexing과 마찬가지로 copy.
c[0] = 100

a[a < 0] = 0 # 음수인 element를 0으로 바꿔준다.
print(a)
```

```python
# operations
a = np.random.randint(1, 10, (5,))
b = np.random.randint(1, 10, (5,))

print(a)
print(b)

print(a + b)
# element-wise multiplication
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
print(a ** b)
```

```python
# operations: broadcasting
a = np.arange(10)
print(a)
b = a + 10 # shape이 다르기 때문에 원칙적으로는 연산 불가.
           # 10을 a의 shape과 맞춰준 다음에 각각 더한다.
           # 이런 과정을 broadcasting이라고 한다.
print(b)
```

```python
# lab: 벡터연산
def softmax(z):
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z) # element-wise operation

logits = np.array([2.0, 1.0, -1.0])
probs = softmax(logits)
print(probs)
print(np.sum(probs))
```

# 2D, 3D, array (rank2(matrix), rank3 tensor)

```python
a = np.random.randint(0, 20, size=(3, 4))
print(a)
print(a.shape)
print(a.ndim)

# 행 단위 이동 -> axis 0 따라 이동
# 열 단위 이동 -> axis 1 따라 이동
```

```python
# indexing
a = np.random.randint(0, 20, size=(4, 5))

print('a')
print(a)
# print(a[0, 0]) # axis0 먼저 인덱싱
# print(a[2, 3])

# print(a[:3, 2:3]) # matlab이랑 거의 똑같음

# b = a[::2, ::2] # not copy, but view
# print('b')
# print(b)

# b[0, 0] = 100
# print('a')
# print(a)

c = a[:, [0, -1]] # fancy indexing. copy
                  # 모든 행, 첫번째, 그리고 마지막 열
c[0, 0] = 100
print('c')
print(c)

print('a')
print(a)
```

```python
# reshaping
a = np.arange(20) # generates a 1d array
print('a')
print(a)

b = a.reshape((4, -1)) # 하나만 쓰고 나머지는 -1 쓰면 자동으로 값을 채워줌.
                       # 이것도 view만 바꾼다.
b[1, 1] = 1000
print('b')
print(b)

print('a')
print(a)

c = a.reshape((4, -1), copy=True) # copy
c[1, 1] = 2000
print('c')
print(c)

print('a')
print(a)
```

```python
a = np.arange(10)
print('a')
print(a)

# b = a[:, None] # 뒤에 axis 추가
# print('b')
# print(b)
# print(b.shape)

c = a[None, :] # 앞에 axis 추가
print('c')
print(c)
print(c.shape)

# # 2차원 공간 안에 1차원 데이터를 둔 것이라 생각하면 됨.

# d = a[np.newaxis, :, np.newaxis]
# print('d')
# print(d)
# print(d.shape)

e = c.reshape((10,))
print(e)
print(e.shape)

e = c.reshape(-1) # 강제로 1차원으로 나열해버린다. returns a view
print(e)
print(e.shape)

e = c.flatten() # returns a copy
print(e)
print(e.shape)
```

```python
# transpose
a = np.arange(10).reshape((5, -1)) # 5x2
print(a)
print(a.T) # transpose. axis를 뒤바꾼다. 2x5
```

```python
# broadcasting
a = np.arange(12).reshape((3,-1)) # (3, 4)
print('a')
print(a)
print(a.shape)

# b = a * 10 # shape이 달라서 연산 불가. 그러나 broadcasting으로 shape을 동일하게 맞춰준다.
# print('b')
# print(b)
# print(b.shape)

# c = np.array([2, 2, 2, 2]) # (4, )
# print('c')
# print(c.shape)

# d = a * c # multiply several columns at once
# print('d')
# print(d)
# print(d.shape)

# shape를 뒤에서부터 맞춰나간다.
# (3, 4), (3, 1)
# -> axis0이 같으니 axis1(4/1) -> 4번 반복해야 함을 알 수 있음.
e = np.array([2, 2, 2]).reshape((3, -1))
print(e.shape)
f = a * e
print('f')
print(f)

# 실제로 복사가 일어나는 것이 아니라 반복하는 것.
```

```python
# sum(), mean(), min(), max(), argmin(), argmax()
np.random.seed(1)
a = np.random.randint(20, size=(4, 5))
print(a)

print(f'sum={a.sum()}')
print(f'mean={a.mean()}')
print(f'min={a.min()}')
print(f'max={a.max()}')

print(f'sum of each row={a.sum(axis=1)}')
print(f'sum of each col={a.sum(axis=0)}')
print(f'mean of each row={a.mean(axis=1)}')
print(f'mean of each col={a.mean(axis=0)}')

b = a.sum(axis=0)

print('b')
print(b)
print(b.shape) # axis가 1개로 줄어버렸다.

b = a.sum(axis=0, keepdims=True)
print('b')
print(b)
print(b.shape)
```

```python
# sorting
a = np.random.randint(1, 20, size=(4, 5))
print('a')
print(a)

# in-place sort
a.sort(axis=0)
print('a.sort(axis=0)')
print(a)

print('b = np.sort(a, axis=0)')
b = np.sort(a, axis=0) # returns a sorted copy of the original array
print(b)
```
