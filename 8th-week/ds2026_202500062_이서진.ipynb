```python
# 1
import numpy as np

a = np.arange(0, 26).reshape((2,-1))
print(a)
```

```python
# 2
a_transposed = a.T
a_reshaped = a.reshape((13,-1))

print(a_transposed)
print(a_reshaped)

# reshape는 각 element들이 메모리상 위치하는 순서는 유지한 채로 array의 모양만 바꾸고,
# transpose는 행과 열을 서로 교환하여 실제 위치가 바뀐다.
```

```python
# 8
a = np.random.rand(4)
print(a)

b = np.random.randn(12).reshape((4,-1))
print(b)

c = np.full((1, 4), 42.)
print(c)
```

```python
# 9
print(f'a={a.shape}')
print(f'b={b.shape}')
print(f'c={c.shape}')

# a는 b와 연산이 불가하고, c를 transpose하면 b와 연산이 가능하다.
# elementwise 연산을 할 때 shape이 맞지 않는 경우 broadcasting이 일어나는데,
# broadcasting이 가능한 조건이, 서로 크기가 맞는 axis가 있어야 한다는 것이다. (튜플 뒤에서부터 확인.)
# c를 transpose하면 그 shape은 (4, 1)이 되고, 여러 행에 한번에 연산이 가능해진다.
```

```python
# 10
# axis2를 일치하게 하려면 b를 transpose시키는 방법밖에 없었다.
print(f'b + a = {b.T + a}')
# 9번에서 설명한 이유로 c를 Transpose시켰다.
# 그러나 b.T + c도 가능하다. 결과의 shape은 각각 (4, 3), (3, 4)로 다르다.
print(f'b + c = {b + c.T}')
```

```python
# 20
a = np.array([[4, 1],
     [2, 4]])
b = np.array([[5, 6],
     [1, 8]])

print((a @ b) @ (b @ a))
```

```python
# 21
c = np.array([[1, 2],
     [8, 4],
     [3, 9]])

# a.shape = (2, 2)
# b.shape(2, 2)
# c.shape = (3, 2)
# 2x2 행렬과 3x2 행렬의 곱셈은 불가능하다. mxn, rxp 행렬의 곱셈은 n=r일 때라야 가능하다.
# [3x2] [2x2] 는 가능하다.
print((c @ a) + (c @ b))
```

```python
# 22
k = np.ones((3, 4)) * np.linspace(-1, 2.5, 4)
print(k)
print(f'올림 : {np.ceil(k)}')
print(f'내림 : {np.floor(k)}')
print(f'반올림 : {np.round(k)}')
```

```python
# 23
# np.sum은 모든 요소의 합을 반환한다.
print(np.sum(k))
# np.cumsum은 각 위치에서의 누적합을 반환한다.
print(np.cumsum(k))
```
