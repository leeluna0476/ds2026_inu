## String Slicing

```python
print(city[:3])
print(city[3:])
print(city[:-2])
print(city[-2:])
print(city[0:3]) # [a:b)
print(city[:])
print(city[0::2])
print(city[::-1])
```

```python
'a' < 'b' # 사전순으로 비교
'10' > '9' # '9' > '1'
```

```python
print('ppp' in 'apple')
print('pp' in 'apple') # strstr in C
```

```python
for l in 'hello':
    print(l)
```

```python
# Remove Vowels
s = input('Enter a string: ')
vowels = 'aeiou'
for c in s:
    if c not in vowels:
        print(c, end='')
print()
```
