# Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Prepare data
X = np.linspace(0, 4*np.pi, 100)
Y_sin = np.sin(X)
Y_cos = np.cos(X)
Y_tan = np.tan(X)

# 큰 값 -> 작은 값 연결을 막기 위해
Y_tan[np.abs(Y_tan) > 5] = np.nan
```

```python
# Render
fig, ax = plt.subplots()
ax.plot(X, Y_sin, label='sin(x)')
ax.plot(X, Y_cos, label='cos(x)')
ax.plot(X, Y_tan, label='tan(x)')

# ax.set_xlabel('x')
# ax.set_ylabel('y')
ax.set(xlabel='x', ylabel='y')
ax.grid(axis='y')
ax.legend(loc='upper right')
```

```python
# lab: plot sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-10, 10, 100)
y = sigmoid(x)

fig, ax = plt.subplots()
ax.plot(x, y, label=r'$sigmoid(x)=\frac{1}{1 + e^{-x}}$', linestyle=':')
ax.set(xlabel='x', ylabel='y')
ax.set_yticks(np.linspace(0, 1, 11))
ax.grid(axis='both')
ax.legend()
plt.show()
```

```python
# scatter plot
# histogram
# 2 axes

# sample을 많이 뽑으면 모집단의 분포와 비슷해짐.

num_points = 100000
x = np.random.randn(num_points)
y = x + 0.1 * np.random.randn(num_points)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].scatter(x, y, edgecolor='k', color='r')
ax[0].set(xlabel='x', ylabel='y')
ax[0].set_title('Scatter plot')
ax[1].hist(x, np.arange(-3, 3, 0.2), edgecolor='k')
ax[1].set(xlabel='x', ylabel='freq')
ax[1].set_title('Histogram')

fig.tight_layout()
```

```python
# lab

def relu(x):
    return np.maximum(0, x)

x = np.linspace(-10, 10, 100)
y1 = sigmoid(x)
y2 = relu(x)

fig, ax = plt.subplots(1, 2, figsize=(15, 5))
ax[0].plot(x, y1)
ax[0].set(xlabel='x', ylabel='y')
ax[0].set_title('y=sigmoid(x)')

ax[1].plot(x, y2)
ax[1].set(xlabel='x', ylabel='y')
ax[1].set_title('y=relu(x)')
```
