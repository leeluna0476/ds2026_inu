```python
a = np.random.randint(-100, 101, size=(10000000,))
# a = -100 + 200 * np.random.rand(10000000)
print(a.min())
print(a.max())
print(a.mean())
print(a.std())
```

```python
a = [[4, 1],
     [2, 4]]
b = [[5, 6],
     [1, 8]]

na = np.array(a)
nb = np.array(b)
print((na @ nb) @ (nb @ na))
```

```python
c = [[1, 2],
     [8, 4],
     [3, 9]]
nc = np.array(c)

print(na.shape)
print(nc.shape)
print(nb.shape)

print((nc @ na) + (nc @ nb))
```

```python
k = np.ones((3, 4)) * np.linspace(-1, 2.5, 4)
print(k)

print(np.ceil(k))
print(np.floor(k))
print(np.round(k))
```

```python
print(np.sum(k,axis=0))
print(np.cumsum(k,axis=0))
```
