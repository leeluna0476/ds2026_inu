```python
FILENAME = 'Alice.txt'
# file을 가리키는 object
reader = open(FILENAME, 'r', encoding='utf-8') # returns a stream (a sequence)
# file system으로부터 reader가 가리키는 파일의 데이터를 메모리로 불러온다.
# object는 그냥 접근하지 못하므로 변수를 통해야 한다.
# 보통 이렇게 한번에 읽어오면 메모리 낭비가 심하므로 여러 번에 나눠서 읽어온다.
data = reader.read() # reads size-bytes or reads until EOF
# 더 이상 파일시스템에 접근할 필요가 없다면 close해야 한다.
# 하나의 파일에 여러 명이 동시에 read/write를 하면 오류가 생길 수 있다.
reader.close()
len(data)
```

```python
FILENAME = 'Alice.txt'
reader = open(FILENAME, 'r', encoding='utf-8')
data = reader.read(64)
while data:
    print(len(data))
    data = reader.read(64)
reader.close()
```

```python
FILENAME = 'Alice.txt'
reader = open(FILENAME, 'r', encoding='utf-8')
data = reader.readline()
while data:
    print(len(data))
    data = reader.readline()
reader.close()
```

```python
FILENAME = 'Alice.txt'
reader = open(FILENAME, 'r', encoding='utf-8')
data = reader.readline()

num_chars, num_words, num_lines = 0, 0, 0
while data:
    num_chars += len(data)
    num_words += len(data.split())
    num_lines += 1
    data = reader.readline()
reader.close()

print(f'average: {num_chars / num_lines:.2f} chars per line')
print(f'average: {num_words / num_lines:.2f} words per line')
```

```python
# readlines: list of lines
FILENAME = 'Alice.txt'
reader = open(FILENAME, 'r', encoding='utf-8')
data = reader.readlines()
print(len(data))
print(data[0])
reader.close()
```

```python
# A file is iterable.
FILENAME = 'Alice.txt'
reader = open(FILENAME, 'r', encoding='utf-8')

num_chars, num_words, num_lines = 0, 0, 0
for line in reader:
    num_chars += len(line)
    num_words += len(line.split())
    num_lines += 1
reader.close()

print(f'average: {num_chars / num_lines:.2f} chars per line')
print(f'average: {num_words / num_lines:.2f} words per line')
```

```python
# write to a file
FILENAME = 'myfile.txt'
writer = open(FILENAME, 'w', encoding='utf-8')

writer.write('hello world\n')
writer.write('I am seojilee\n')

l = ['hello jina\n', 'nice to meet you\n', 'i am hungry professor\n']
writer.writelines(l)

writer.close()
```

```python
# append to a file
FILENAME = 'myfile.txt'
writer = open(FILENAME, 'a', encoding='utf-8')

writer.write('Prof. Wchkang\n')
writer.write('I am seojilee\n')

l = ['I feel so hungry\n', 'Please finish the class\n']
writer.writelines(l)

writer.close()
```

```python
# with statement
# file 매번 close하는 게 매우 귀찮다.
# with는 그걸 자동으로 해준다.
FILENAME = 'myfile.txt'
with open(FILENAME, 'r', encoding='utf-8') as bong:
    for line in bong:
        print(line, end='')
# with를 벗어나면 자동으로 close된다.
```

```python
# lab: copy a file
SRC = 'alice.txt'
DST = 'alice_copied.txt'
src = open(SRC, 'r', encoding='utf-8')
dst = open(DST, 'w', encoding='utf-8')

for line in src:
    dst.write(line)

src.close()
dst.close()
```

```python
# lab: copy a file
SRC = 'alice.txt'
DST = 'alice_copied.txt'
with open(SRC, 'r', encoding='utf-8') as src:
    with open(DST, 'w', encoding='utf-8') as dst:
        for line in src:
            dst.write(line)
```
