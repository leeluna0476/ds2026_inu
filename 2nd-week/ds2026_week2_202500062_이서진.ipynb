```python
# 13
n = int(input('Enter a number: (n >= 0)'))

print('You enetered:', n)
f = 1
i = 1
while i <= n:
    f *= i
    i += 1

print('factorial:', f)
```

```python
# 14
n = 5

for i in range(n):
    for j in range(n * 2):
        if n - i <= j <= n + i and i % 2 != j % 2:
            print('*', end='')
        else:
            print(' ', end='')
    print()
```

```python
# 16
mail = input('Enter String(8-alphabet) : ')
count = 0

for c in mail:
    if c == 'M' or c == 'S':
        count += 1

if count >= 5:
    print('Secret mail!')
else:
    print('평범한 문자!')
```
