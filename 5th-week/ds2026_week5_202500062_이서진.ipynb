```python
# 6

def recurse(num, s, Roman):
    for i in range(len(Roman) - 1, -1, -1):
        if Roman[i][0] - 1 <= num:
            break
    diff = num - Roman[i][0]
    if diff < 0:
        s = recurse(-diff, s, Roman)
        s += Roman[i][1]
    elif diff > 0:
        s += Roman[i][1]
        s = recurse(diff, s, Roman)
    else:
        s += Roman[i][1]
    return s

def Roma(num):
    Roman = [(1, 'I'), (5, 'V'), (10, 'X')]
    return recurse(num, '', Roman)
```

```python
# 7
R_num = [Roma(n) for n in range(1, 11)]
print(R_num)
```
