## alias vs (shallow) copy vs deep copy

```python
# b = a.copy()
b = a[:] # list를 새로 생성.
print(id(b))
print(a[0] is b[0])
b[0] = 1
print(b)
print(a)
# b[0] 과 a[0]이 가리키는 object는 동일하다.
# b[0] = 1 -> b[0]이 100 대신 1을 가리키도록 변경한다. a[0]은 여전히 1을 가리킨다.
```

```python
a = [1, [2, 3]]
b = a
b[1][0] = 4 # b[1]과 a[1]은 같은 object(list) 가리키고, 이는 mutable이므로 a[1][0]도 같이 바뀐다
print(a)
```

```python
import copy
b = copy.deepcopy(a) # 모든 mutable object를 복사한다.
b[1][0] = 5
print(a) # a 더 이상 영향을 받지 않는다.
```

```python
# list as arguments

def double(mylist):
    for index, value in enumerate(mylist):
        mylist[index] = 2 * value

things = [1, 2, 'hello']
double(things)
print(things)
```

```python
double(things[:]) # copy. 원본은 변하지 않는다.
print(things)
double(things)
print(things)
```

# Tuple
- immutable sequence of objects

```python
fruits = [ 'apple', 'kiwi', 'banana', 'grape', 'mango' ]

print(fruits)
```

```python
#fruits = { 'apple', 'kiwi', 'banana', 'grape', 'mango' } # 소괄호는 set이다
fruits = ( 'apple', 'kiwi', 'banana', 'grape', 'mango' )
print(type(fruits))
print(fruits[:3])
print(fruits[3:])
print(fruits[-1])

for f in fruits:
    print(f)
```

```python
fruits[0] = 'strawberry' # immutable. error!
```

- 바뀔 가능성이 없다면 캐시에 올려두고 쓰기 편리하다.
- 멀티코어 컴퓨터에서 서로 메모리의 값을 바꿀 때...
- 어차피 바뀔 일 없으니까 각자 캐시에 가져다가 읽을 때 쓰면 된다.
- mutable이라면 매번 메모리를 읽어야 한다

```python
gases = 'a', 'b', 'c' # assign a tuple
# gases = ('a' 'b', 'c')
gases
```

```python
x, y, z = gases
x, y, _ = gases # 안 쓰는 건 _로 버린다
print(x, y)
```

```python
t = ('a', 'b', [1, 2])
print(t)
t[2][0] = 42 # the tuple is immutable, but the list object in the tuple is mutable
print(t)
```

## List comprehension

```python
# pythonic
# [expression for variable in iterable]
[i * 2 for i in range(10)]
```

```python
[ord(c) for c in 'bong']
```

```python
li = [('a', 1), ('c', 3), ('d', 2)]

# list is iterable
# each element(tuple) is unpacked to a str and an int
li2 = [n for _, n in li]
li2
```

```python
num_string = '1 2 3 4 5'
num_list = [int(c) for c in num_string.split()]
num_list
```

```python
num_pairs = [(1, 2), (3, 4), (5, 6)]
num_exchanged = [(b, a) for (a, b) in num_pairs]
num_exchanged
```

```python
num_string = '1, 2, 3, 4, 5, 6'
print([int(c) for c in num_string.split(', ')])
```

```python
num_pairs = [(1, 2), (3, 4), (6, 5)]
num_exchanged = [(b, a) for (a, b) in num_pairs if a < b]
num_exchanged
```

```python
li = [3, 6, 2, 7, 1, 9]
# 짝수만 골라내어 제곱한다
[n**2 for n in li if n % 2 == 0]
```

```python
num_pairs = [(1, 2), (3, 4), (6, 5)]
# else를 포함하려면 아래 형식을 따라야 한다
# [expr1 if condition else expr2 for var in iterable]
num_exchanged = [(b, a) if a < b else (a, b) for (a, b) in num_pairs]
num_exchanged
```

```python
li = [3, 6, 2, 7, 1, 9]
# 짝수는 남기고 홀수는 0으로
[n**2 if n % 2 == 0 else 0 for n in li]
```
